## 실험/분석 사용 순서

- 시뮬레이션 조건 설정 → **RECALCULATE**
- 실측 전극 정보, 간격, 전압, 전류, 풍속 입력
- **SAVE EXPERIMENT** → `experiment_results.csv`에 누적
- **SHOW COMPARISON** → 실측/시뮬레이션 비교
- **ANALYZE DATA** → 간격 1 mm별 통계, 형상별 평균±SD, 오차 분석, 최적 관측 조건
- **EXPORT SUMMARY** → `gap_summary_1mm.csv`, `shape_summary.csv` 다운로드
- **DOWNLOAD CSV**로 원본 실험 CSV 백업, 다음 세션에서 **UPLOAD CSV**로 이어서 사용

주의: 최적 조건 추천은 현재 CSV에 실제로 들어있는 관측 행 중 최선값을 선택하는 기능이며, 물리적 전역 최적점을 외삽하는 기능은 아닙니다.
