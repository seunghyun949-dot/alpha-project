## 실험 기록 사용 순서

- 시뮬레이션 조건을 설정하고 **RECALCULATE** 실행
- Emitter type / diameter / Collector type / gap 입력
- 실측 전압(kV), 전류(mA), 풍속(m/s) 입력
- **SAVE EXPERIMENT** → `experiment_results.csv`에 한 행 추가
- **SHOW COMPARISON** → 실측/시뮬레이션 비교 그래프 표시
- **DOWNLOAD CSV** → 현재 실험 데이터 백업
- 다음 Colab 세션에서 기존 데이터를 계속 쓰려면 **UPLOAD CSV** 사용

주의: 현재 시뮬레이션 출력은 검증된 절대 추력 예측값이 아니라 상대 비교용 모델값입니다.
